```python
# =======================
# IMPORT LIBRARIES
# =======================
from flask import Flask, request
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler

# =======================
# FLASK APP
# =======================
app = Flask(__name__)

# =======================
# SAMPLE DATA (EXAM-FRIENDLY)
# Example: Binary classification
# Features: [Hours_Studied, Attendance]
# Label: Pass (0 / 1)
# =======================
X = np.array([
    [2, 60],
    [3, 65],
    [4, 70],
    [5, 75],
    [6, 80],
    [7, 85]
])
y = np.array([0, 0, 0, 1, 1, 1])

# =======================
# FEATURE SCALING
# =======================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# =======================
# TRAIN ANN MODEL
# =======================
model = MLPClassifier(
    hidden_layer_sizes=(10, 10),
    activation='relu',
    max_iter=1000,
    random_state=42
)
model.fit(X_scaled, y)

# =======================
# HOME ROUTE
# =======================
@app.route('/')
def home():
    return '''
    <h2>ANN Prediction</h2>
    <form action="/predict" method="post">
        <label>Hours Studied:</label><br>
        <input type="number" name="hours"><br><br>

        <label>Attendance (%):</label><br>
        <input type="number" name="attendance"><br><br>

        <input type="submit" value="Predict">
    </form>
    '''

# =======================
# PREDICTION ROUTE
# =======================
@app.route('/predict', methods=['POST'])
def predict():
    hours = float(request.form['hours'])
    attendance = float(request.form['attendance'])

    input_data = np.array([[hours, attendance]])
    input_scaled = scaler.transform(input_data)

    prediction = model.predict(input_scaled)

    if prediction[0] == 1:
        result = "Pass"
    else:
        result = "Fail"

    return f"<h3>Prediction Result: {result}</h3>"

# =======================
# RUN FLASK APP
# =======================
if __name__ == '__main__':
    app.run(debug=True)
```
