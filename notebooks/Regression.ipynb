```python
# =======================
# IMPORT LIBRARIES
# =======================
from flask import Flask, request
import numpy as np
from sklearn.linear_model import LinearRegression

# =======================
# FLASK APP
# =======================
app = Flask(__name__)

# =======================
# DATA FOR LINEAR REGRESSION
# =======================
# Example: House Size vs Price
X = np.array([[500], [1000], [1500], [2000], [2500]])
y = np.array([100000, 200000, 300000, 400000, 500000])

# =======================
# TRAIN LINEAR REGRESSION MODEL
# =======================
model = LinearRegression()
model.fit(X, y)

# =======================
# HOME ROUTE
# =======================
@app.route('/')
def home():
    return '''
    <h2>Linear Regression Prediction</h2>
    <form action="/predict" method="post">
        <label>Enter House Size (sq ft):</label><br><br>
        <input type="number" name="size"><br><br>
        <input type="submit" value="Predict Price">
    </form>
    '''

# =======================
# PREDICTION ROUTE
# =======================
@app.route('/predict', methods=['POST'])
def predict():
    size = float(request.form['size'])
    prediction = model.predict([[size]])

    return f"<h3>Predicted House Price: {prediction[0]}</h3>"

# =======================
# RUN FLASK APP
# =======================
if __name__ == '__main__':
    app.run(debug=True)
```
